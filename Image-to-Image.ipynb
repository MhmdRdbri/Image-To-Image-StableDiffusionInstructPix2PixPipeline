```python
!pip install diffusers
!pip install torch
!pip install pillow
!pip install torch==2.0+cu118


from diffusers import StableDiffusionInstructPix2PixPipeline, EulerAncestralDiscreteScheduler
import torch
from PIL import Image

pipe = StableDiffusionInstructPix2PixPipeline.from_pretrained(
    "timbrooks/instruct-pix2pix", torch_dtype=torch.float16, safety_checker=None
)
pipe.scheduler = EulerAncestralDiscreteScheduler.from_config(pipe.scheduler.config)
pipe.to("cuda")

image = Image.open("images.jpg").convert("RGB")
prompt = "Change only the kitchen cabinets to a modern metal style, such as brushed steel or matte metal. Keep the cabinet shapes, sizes, and positions exactly the same. Do not alter the stove, oven, microwave, countertops, sink, faucet, backsplash, flooring, ceiling, lighting, or any other objects in the image. Preserve the original lighting, reflections, and overall photorealism of the scene. All other elements must remain completely unchanged."

edited_image = pipe(
    prompt=prompt,
    image=image,
    num_inference_steps=50,
    image_guidance_scale=1.5
).images[0]

edited_image.save("result.png")
```

```python
!pip install --force-reinstall numpy
!pip install -q realesrgan gfpgan
!pip install -q git+https://github.com/xinntao/Real-ESRGAN.git
!pip install torch==2.1.0 torchvision==0.16.0 --upgrade
import os
import torch
from realesrgan import RealESRGAN
from PIL import Image
import requests
import subprocess

# Load image from URL
input_path = 'https://raw.githubusercontent.com/xinntao/Real-ESRGAN/master/inputs/lena.png'
img = Image.open(requests.get(input_path, stream=True).raw).convert('RGB')

# Set device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)

# Initialize model
model = RealESRGAN(device, scale=4)

# Download model weights if needed
model_path = os.path.expanduser('~/.cache/realesrgan/RealESRGAN_x4.pth')
if not os.path.exists(model_path):
    os.makedirs(os.path.dirname(model_path), exist_ok=True)
    subprocess.run([
        "wget", "-O", model_path,
        "https://github.com/xinntao/Real-ESRGAN/releases/download/v0.2.5/RealESRGAN_x4.pth"
    ])
# Load weights
model.load_weights(model_path)

# Super-resolve
sr_image = model.predict(img)

# Save and show result
sr_image.save('output.png')
sr_image.show()
```
